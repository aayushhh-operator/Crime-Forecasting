```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# Load the dataset
file_path = '/content/Clean_LA_Crime.csv'
data = pd.read_csv(file_path)

# Parse datetime columns if they are not already parsed
data['DATE OCC'] = pd.to_datetime(data['DATE OCC'])
data['TIME OCC'] = pd.to_datetime(data['TIME OCC'], format='%H:%M:%S')

# Feature Engineering: Adding time-related features
data['DayOfWeek'] = data['DATE OCC'].dt.dayofweek
data['DayOfYear'] = data['DATE OCC'].dt.dayofyear
data['Hour'] = data['TIME OCC'].dt.hour
data['Month'] = data['DATE OCC'].dt.month
data['Year'] = data['DATE OCC'].dt.year

# Encode categorical columns
label_encoder = LabelEncoder()
data['CrmCdDescEncoded'] = label_encoder.fit_transform(data['Crm Cd Desc'])

# Create a target variable for crime counts
data['CrimeCount'] = 1

# Add Lag features and Rolling Mean features
def add_lags(df, lag=1):
    df['CrimeLag'] = df['CrimeCount'].shift(lag)  # Lag of previous crime occurrence
    df['CrimeRollingMean'] = df['CrimeCount'].rolling(window=lag).mean()  # Rolling mean of previous occurrences
    return df

# Aggregating Data
data['DateHour'] = data['DATE OCC'].dt.floor('H')
hourly_counts = data.groupby('DateHour').size()

# Add lag and rolling mean features
hourly_features = data.groupby('DateHour').agg({
    'Hour': 'first',
    'DayOfWeek': 'first',
    'Month': 'first',
    'Year': 'first',
    'CrmCdDescEncoded': 'mean',
}).reset_index()

hourly_features['CrimeCount'] = hourly_counts.values
hourly_features = add_lags(hourly_features, lag=3)
hourly_features = hourly_features.dropna()

# Normalize the features
scaler = MinMaxScaler()
feature_scaler = MinMaxScaler()
target_scaler = MinMaxScaler()

X_hourly = feature_scaler.fit_transform(hourly_features.drop(['DateHour', 'CrimeCount'], axis=1).values)
y_hourly = target_scaler.fit_transform(hourly_features['CrimeCount'].values.reshape(-1, 1))

# Train-test split (chronologically)
split_point = int(len(hourly_features) * 0.8)
X_train_hourly, X_test_hourly = X_hourly[:split_point], X_hourly[split_point:]
y_train_hourly, y_test_hourly = y_hourly[:split_point], y_hourly[split_point:]

# Reshape for LSTM
X_train_hourly_lstm = np.reshape(X_train_hourly, (X_train_hourly.shape[0], X_train_hourly.shape[1], 1))
X_test_hourly_lstm = np.reshape(X_test_hourly, (X_test_hourly.shape[0], X_test_hourly.shape[1], 1))

# Define the LSTM model with Bidirectional layers and EarlyStopping
def build_lstm_model(optimizer='adam', units=50, dropout_rate=0.2):
    model = Sequential()
    model.add(Bidirectional(LSTM(units=units, return_sequences=True), input_shape=(X_train_hourly_lstm.shape[1], 1)))
    model.add(Dropout(dropout_rate))
    model.add(LSTM(units=units, return_sequences=False))
    model.add(Dropout(dropout_rate))
    model.add(Dense(1))
    model.compile(optimizer=optimizer, loss='mean_squared_error')
    return model

# EarlyStopping callback to avoid overfitting
early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
lr_reduction = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=1e-6)

# Hyperparameter combinations
param_combinations = [
    {'optimizer': 'adam', 'units': 50, 'dropout_rate': 0.2},
    {'optimizer': 'rmsprop', 'units': 50, 'dropout_rate': 0.2},
    {'optimizer': 'adam', 'units': 100, 'dropout_rate': 0.3},
    {'optimizer': 'nadam', 'units': 100, 'dropout_rate': 0.3},
]

best_mse = float('inf')
best_params = None

# Hyperparameter tuning loop
for params in param_combinations:
    print(f"Training with params: {params}")
    model = build_lstm_model(optimizer=params['optimizer'], units=params['units'], dropout_rate=params['dropout_rate'])
    model.fit(X_train_hourly_lstm, y_train_hourly, epochs=50, batch_size=32, validation_data=(X_test_hourly_lstm, y_test_hourly),
              callbacks=[early_stopping, lr_reduction], verbose=1)

    # Evaluate the model on the test set
    y_pred = model.predict(X_test_hourly_lstm)
    mse = mean_squared_error(y_test_hourly, y_pred)
    print(f'Mean Squared Error: {mse}')

    if mse < best_mse:
        best_mse = mse
        best_params = params

# Best parameters found
print(f"Best parameters: {best_params}")
print(f"Best MSE: {best_mse}")

# Train the best model
best_model = build_lstm_model(optimizer=best_params['optimizer'], units=best_params['units'], dropout_rate=best_params['dropout_rate'])
best_model.fit(X_train_hourly_lstm, y_train_hourly, epochs=50, batch_size=32, validation_data=(X_test_hourly_lstm, y_test_hourly), verbose=0)

# Make predictions
y_pred_best = best_model.predict(X_test_hourly_lstm)

# Inverse transform the predictions and actual values
y_pred_best_inv = target_scaler.inverse_transform(y_pred_best)
y_test_hourly_inv = target_scaler.inverse_transform(y_test_hourly)

# Plot the actual vs predicted values
plt.figure(figsize=(12, 6))
plt.plot(y_test_hourly_inv, label="Actual")
plt.plot(y_pred_best_inv, label="Predicted")
plt.title("Hourly Crime Predictions (Actual vs Predicted)")
plt.xlabel("Time Steps")
plt.ylabel("Crime Count")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# Load the dataset
file_path = '/content/Clean_LA_Crime.csv'
data = pd.read_csv(file_path)

# Parse datetime columns if they are not already parsed
data['DATE OCC'] = pd.to_datetime(data['DATE OCC'])
data['TIME OCC'] = pd.to_datetime(data['TIME OCC'], format='%H:%M:%S')

# Feature Engineering: Adding time-related features
data['DayOfWeek'] = data['DATE OCC'].dt.dayofweek
data['DayOfYear'] = data['DATE OCC'].dt.dayofyear
data['Hour'] = data['TIME OCC'].dt.hour
data['Month'] = data['DATE OCC'].dt.month
data['Year'] = data['DATE OCC'].dt.year

# Encode categorical columns
label_encoder = LabelEncoder()
data['CrmCdDescEncoded'] = label_encoder.fit_transform(data['Crm Cd Desc'])

# Create a target variable for crime counts
data['CrimeCount'] = 1

# Add Lag features and Rolling Mean features
def add_lags(df, lag=1):
    df['CrimeLag'] = df['CrimeCount'].shift(lag)  # Lag of previous crime occurrence
    df['CrimeRollingMean'] = df['CrimeCount'].rolling(window=lag).mean()  # Rolling mean of previous occurrences
    return df

# Aggregating Data for hourly, daily, and monthly
data['DateHour'] = data['DATE OCC'].dt.floor('H')
data['Date'] = data['DATE OCC'].dt.date
data['YearMonth'] = data['DATE OCC'].dt.to_period('M')

hourly_counts = data.groupby('DateHour').size()
daily_counts = data.groupby('Date').size()
monthly_counts = data.groupby('YearMonth').size()

# Feature aggregation for hourly, daily, and monthly
def aggregate_features(df, period='hour'):
    if period == 'hour':
        return df.groupby('DateHour').agg({
            'Hour': 'first',
            'DayOfWeek': 'first',
            'Month': 'first',
            'Year': 'first',
            'CrmCdDescEncoded': 'mean',
        }).reset_index()
    elif period == 'day':
        return df.groupby('Date').agg({
            'DayOfWeek': 'first',
            'Month': 'first',
            'Year': 'first',
            'CrmCdDescEncoded': 'mean',
        }).reset_index()
    elif period == 'month':
        return df.groupby('YearMonth').agg({
            'Month': 'first',
            'Year': 'first',
            'CrmCdDescEncoded': 'mean',
        }).reset_index()

hourly_features = aggregate_features(data, period='hour')
daily_features = aggregate_features(data, period='day')
monthly_features = aggregate_features(data, period='month')

# Add lag features
hourly_features['CrimeCount'] = hourly_counts.values
daily_features['CrimeCount'] = daily_counts.values
monthly_features['CrimeCount'] = monthly_counts.values

hourly_features = add_lags(hourly_features, lag=3)
daily_features = add_lags(daily_features, lag=3)
monthly_features = add_lags(monthly_features, lag=3)

hourly_features = hourly_features.dropna()
daily_features = daily_features.dropna()
monthly_features = monthly_features.dropna()

# Normalize the features and target
scaler = MinMaxScaler()
feature_scaler = MinMaxScaler()
target_scaler = MinMaxScaler()

X_hourly = feature_scaler.fit_transform(hourly_features.drop(['DateHour', 'CrimeCount'], axis=1).values)
y_hourly = target_scaler.fit_transform(hourly_features['CrimeCount'].values.reshape(-1, 1))

X_daily = feature_scaler.fit_transform(daily_features.drop(['Date', 'CrimeCount'], axis=1).values)
y_daily = target_scaler.fit_transform(daily_features['CrimeCount'].values.reshape(-1, 1))

X_monthly = feature_scaler.fit_transform(monthly_features.drop(['YearMonth', 'CrimeCount'], axis=1).values)
y_monthly = target_scaler.fit_transform(monthly_features['CrimeCount'].values.reshape(-1, 1))

# Train-test split (chronologically)
split_point_hourly = int(len(hourly_features) * 0.8)
split_point_daily = int(len(daily_features) * 0.8)
split_point_monthly = int(len(monthly_features) * 0.8)

X_train_hourly, X_test_hourly = X_hourly[:split_point_hourly], X_hourly[split_point_hourly:]
y_train_hourly, y_test_hourly = y_hourly[:split_point_hourly], y_hourly[split_point_hourly:]

X_train_daily, X_test_daily = X_daily[:split_point_daily], X_daily[split_point_daily:]
y_train_daily, y_test_daily = y_daily[:split_point_daily], y_daily[split_point_daily:]

X_train_monthly, X_test_monthly = X_monthly[:split_point_monthly], X_monthly[split_point_monthly:]
y_train_monthly, y_test_monthly = y_monthly[:split_point_monthly], y_monthly[split_point_monthly:]

# Reshape for LSTM
X_train_hourly_lstm = np.reshape(X_train_hourly, (X_train_hourly.shape[0], X_train_hourly.shape[1], 1))
X_test_hourly_lstm = np.reshape(X_test_hourly, (X_test_hourly.shape[0], X_test_hourly.shape[1], 1))

X_train_daily_lstm = np.reshape(X_train_daily, (X_train_daily.shape[0], X_train_daily.shape[1], 1))
X_test_daily_lstm = np.reshape(X_test_daily, (X_test_daily.shape[0], X_test_daily.shape[1], 1))

X_train_monthly_lstm = np.reshape(X_train_monthly, (X_train_monthly.shape[0], X_train_monthly.shape[1], 1))
X_test_monthly_lstm = np.reshape(X_test_monthly, (X_test_monthly.shape[0], X_test_monthly.shape[1], 1))

# Define the LSTM model with Bidirectional layers and EarlyStopping
def build_lstm_model(optimizer='adam', units=50, dropout_rate=0.2):
    model = Sequential()
    model.add(Bidirectional(LSTM(units=units, return_sequences=True), input_shape=(X_train_hourly_lstm.shape[1], 1)))
    model.add(Dropout(dropout_rate))
    model.add(LSTM(units=units, return_sequences=False))
    model.add(Dropout(dropout_rate))
    model.add(Dense(1))
    model.compile(optimizer=optimizer, loss='mean_squared_error')
    return model

# EarlyStopping callback to avoid overfitting
early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
lr_reduction = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=1e-6)

# Train the best model for each period (hourly, daily, and monthly)
def train_lstm_model(X_train_lstm, y_train, X_test_lstm, y_test, period):
    model = build_lstm_model(optimizer='adam', units=50, dropout_rate=0.2)
    model.fit(X_train_lstm, y_train, epochs=50, batch_size=32, validation_data=(X_test_lstm, y_test),
              callbacks=[early_stopping, lr_reduction], verbose=1)

    y_pred = model.predict(X_test_lstm)
    mse = mean_squared_error(y_test, y_pred)
    print(f'{period.capitalize()} - Test MSE: {mse:.4f}')

    # Inverse transform for better visualization
    y_pred_inv = target_scaler.inverse_transform(y_pred)
    y_test_inv = target_scaler.inverse_transform(y_test)

    plt.figure(figsize=(12, 6))
    plt.plot(y_test_inv, label="Actual")
    plt.plot(y_pred_inv, label="Predicted")
    plt.title(f'{period.capitalize()} Crime Predictions (Actual vs Predicted)')
    plt.xlabel("Time Steps")
    plt.ylabel("Crime Count")
    plt.legend()
    plt.grid(True)
    plt.show()

    return mse

# Evaluate the models for Hourly, Daily, and Monthly Data
hourly_mse = train_lstm_model(X_train_hourly_lstm, y_train_hourly, X_test_hourly_lstm, y_test_hourly, 'hour')
daily_mse = train_lstm_model(X_train_daily_lstm, y_train_daily, X_test_daily_lstm, y_test_daily, 'day')
monthly_mse = train_lstm_model(X_train_monthly_lstm, y_train_monthly, X_test_monthly_lstm, y_test_monthly, 'month')

# Print RMSE values separately
hourly_rmse = np.sqrt(hourly_mse)
daily_rmse = np.sqrt(daily_mse)
monthly_rmse = np.sqrt(monthly_mse)

print(f"Hourly RMSE: {hourly_rmse:.4f}")
print(f"Daily RMSE: {daily_rmse:.4f}")
print(f"Monthly RMSE: {monthly_rmse:.4f}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam


data['DATE OCC'] = pd.to_datetime(data['DATE OCC'])
data['DateHour'] = data['DATE OCC'].dt.floor('H')
data['DateDay'] = data['DATE OCC'].dt.date
data['DateMonth'] = data['DATE OCC'].dt.month

# Aggregate crime counts by hour, day, and month
hourly_data = data.groupby('DateHour').size()
daily_data = data.groupby('DateDay').size()
monthly_data = data.groupby('DateMonth').size()

# Check the size of monthly data
print(f"Monthly Data Size: {len(monthly_data)}")  # To check if it's smaller than 'look_back'

# Normalize the data
scaler = MinMaxScaler(feature_range=(0, 1))

hourly_data_scaled = scaler.fit_transform(hourly_data.values.reshape(-1, 1))
daily_data_scaled = scaler.fit_transform(daily_data.values.reshape(-1, 1))
monthly_data_scaled = scaler.fit_transform(monthly_data.values.reshape(-1, 1))

# Create datasets for training and testing
def create_dataset(data, look_back=24):
    X, y = [], []
    for i in range(len(data) - look_back):
        X.append(data[i:(i + look_back), 0])
        y.append(data[i + look_back, 0])
    return np.array(X), np.array(y)

# Prepare data for training
look_back = 12  # Adjusted for monthly data

X_hourly, y_hourly = create_dataset(hourly_data_scaled, look_back)
X_daily, y_daily = create_dataset(daily_data_scaled, look_back)
X_monthly, y_monthly = create_dataset(monthly_data_scaled, look_back)

# Check if monthly dataset is too small for the look_back window
if X_monthly.shape[0] == 0:
    print("Not enough data points for monthly predictions with the given look_back value.")
    look_back = 3  # Adjust look_back to 3 months (you can change this based on your dataset)

    # Recreate the dataset with the adjusted look_back
    X_monthly, y_monthly = create_dataset(monthly_data_scaled, look_back)
    if X_monthly.shape[0] == 0:
        raise ValueError("Still not enough data points after adjusting the look_back value.")

# Reshape data for LSTM (samples, time steps, features)
X_hourly = X_hourly.reshape(X_hourly.shape[0], X_hourly.shape[1], 1)
X_daily = X_daily.reshape(X_daily.shape[0], X_daily.shape[1], 1)
X_monthly = X_monthly.reshape(X_monthly.shape[0], X_monthly.shape[1], 1)

# Model definition
def build_model(input_shape):
    model = Sequential()
    model.add(Bidirectional(LSTM(units=50, activation='relu', input_shape=input_shape)))
    model.add(Dropout(0.3))  # Dropout for regularization
    model.add(Dense(units=1))
    model.compile(optimizer=Adam(learning_rate=0.001), loss='mean_squared_error')
    return model

# Early stopping callback
early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

# Train models for Hourly, Daily, and Monthly predictions
def train_and_evaluate(X_train, y_train, X_test, y_test):
    model = build_model((X_train.shape[1], 1))
    history = model.fit(X_train, y_train, epochs=50, batch_size=32, validation_data=(X_test, y_test),
                        callbacks=[early_stopping], verbose=2)

    # Predictions
    predictions = model.predict(X_test)

    # RMSE Calculation
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    return history, predictions, rmse

# Train and evaluate Hourly predictions
history_hourly, predictions_hourly, rmse_hourly = train_and_evaluate(X_hourly, y_hourly, X_hourly, y_hourly)
# Train and evaluate Daily predictions
history_daily, predictions_daily, rmse_daily = train_and_evaluate(X_daily, y_daily, X_daily, y_daily)
# Train and evaluate Monthly predictions
history_monthly, predictions_monthly, rmse_monthly = train_and_evaluate(X_monthly, y_monthly, X_monthly, y_monthly)

# Plotting Hourly, Daily, and Monthly Predictions
def plot_predictions(actual, predicted, title):
    plt.figure(figsize=(14, 7))
    plt.plot(actual, label="Actual")
    plt.plot(predicted, label="Predicted")
    plt.title(title)
    plt.xlabel("Time Steps")
    plt.ylabel("Crime Count")
    plt.legend()
    plt.show()

# Plot for Hourly
plot_predictions(hourly_data_scaled[look_back:], predictions_hourly, "Hour Crime Predictions (Actual vs Predicted)")
# Plot for Daily
plot_predictions(daily_data_scaled[look_back:], predictions_daily, "Day Crime Predictions (Actual vs Predicted)")
# Plot for Monthly
plot_predictions(monthly_data_scaled[look_back:], predictions_monthly, "Month Crime Predictions (Actual vs Predicted)")

# Print RMSE for Hourly, Daily, and Monthly
print(f"Hourly RMSE: {rmse_hourly}")
print(f"Daily RMSE: {rmse_daily}")
print(f"Monthly RMSE: {rmse_monthly}")
```
